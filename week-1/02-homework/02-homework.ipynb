1. Determine the cost of some number of pounds of cement. Assume there is a sales tax that applies.
  

```python
# Exercise 1 - Part 1

pounds_of_cement = 10
dolloar_per_pound = 2
tax_rate = 0.1025  # 10.25% tax rate

subtotal = (pounds_of_cement) * (dolloar_per_pound)
sales_tax = (subtotal) * (tax_rate)
total_cost = (subtotal) + (sales_tax)

print("The total cost is: $" + str(total_cost))
```

2. Given the cost of various containers of eggs (perhaps half a dozen, one dozen, and 1 1/2 dozen), determine the cost _per egg_ for each of those containers.
  

```python
# Exercise 1 - Part 2

#. Given the cost of various containers of eggs
price_half_dozen = 6
price_a_dozen = 10

#. Determine the cost per egg for each
cost_per_egg_half_dozen = (price_half_dozen) / 6
cost_per_egg_a_dozen = (price_a_dozen) / 12

print("Cost per egg for a half dozen: $" + str(cost_per_egg_half_dozen))
print("Cost per egg for a dozen: $" + str(cost_per_egg_a_dozen))
```
